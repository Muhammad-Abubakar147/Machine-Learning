**import library**

```python
import pandas as pd 
```

```python
import requests
```

```python
response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=296948dca878a7f7295a340b20e6a962")
```

**Convert into json**

```python
response.json()["results"]
```

**Converting in dataframe**

```python
df = pd.DataFrame(response.json()["results"])[
    ["id", "title", "overview", "release_date",
     "popularity", "vote_average", "vote_count"]
]

print(df)
```

```python
df.head()
```

**Working with Loop**

```python
df=pd.DataFrame()
print(df)
```

```python
for i in range (1,229):
    response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=296948dca878a7f7295a340b20e6a962".format(i))
    tem_df = pd.DataFrame(response.json()["results"])[
    ["id", "title", "overview", "release_date",
     "popularity", "vote_average", "vote_count"]
]

print(tem_df)
df = pd.concat([df, tem_df], ignore_index=True)
print(df)
```
